```python
# 일단 확인
folder = '/content/other_face_1'

# Extract the names of all txt files in the directory
annotation_files = [f for f in os.listdir(folder) if f.endswith('.txt')]

class_values = []

# Loop over each file
for file in annotation_files:
  filepath = os.path.join(folder, file)

  # Open the file and read the lines
  with open(filepath, 'r') as f:
    lines = f.readlines()

  # For each line, split by space and append the class value (first item) to the list
  for line in lines:
    class_values.append(int(line.split()[0]))

# Check the unique class values
unique_class_values = set(class_values)
print("Unique class values are: ", unique_class_values)
```

```python
# 변경하기 코드

import os

folder = '/content/other_face_1'

# Extract the names of all txt files in the directory
annotation_files = [f for f in os.listdir(folder) if f.endswith('.txt')]

# Loop over each file
for file in annotation_files:
  filepath = os.path.join(folder, file)

  # Open the file and read the lines
  with open(filepath, 'r') as f:
    lines = f.readlines()

  # For each line, split by space and replace the class value (first item) to 0
  new_lines = []
  for line in lines:
    items = line.split()
    items[0] = '0'  # Change class value to 0
    new_lines.append(' '.join(items))

  # Write the new lines back to the file
  with open(filepath, 'w') as f:
    f.write('\n'.join(new_lines))

print("All class labels have been changed to 0.")
```

```python
import glob
import os

label_paths = glob.glob('/content/datasets/labels/train/*.txt') + glob.glob('/content/datasets/labels/val/*.txt')

for label_file in label_paths:
    try:
        with open(label_file, 'r') as file:
            lines = file.readlines()

        # 중복 제거를 위해 set을 사용하고 한놈만 남겨 놓자
        unique_lines = list(set(line.strip() for line in lines))

        # 중복이 제거 후에 파일을 덮어써보자
        if len(unique_lines) != len(lines):
            with open(label_file, 'w') as file:
                file.writelines(line + '\n' for line in unique_lines)
            print(f"중복 라벨 제거 완료: {label_file}")
        else:
            print(f"중복 없음: {label_file}")

    except Exception as e:
        print(f"오류 발생: {label_file} - {e}")
```
