## **2.숫자 --> 범주**

### **(1) 시각화**

* 히스토그램을 Survived로 나눠서 그려봅시다.

```python
sns.histplot(x='Age', data = titanic, hue = 'Survived')
plt.show()
```

* kdeplot을 그려봅시다.  
* 두가지 방법이 있습니다.
    * ① kdeplot(   , hue = 'Survived') 
        * 생존여부의 비율이 유지된 채로 그려짐
        * 두 그래프의 아래 면적의 합이 1
    * ② kdeplot(   , hue = 'Survived', common_norm = False)
        * 생존여부 각각 아래 면적의 합이 1인 그래프
    * ③ kdeplot(   , hue = 'Survived', multiple = 'fill')
        * 나이에 따라 생존여부 **비율**을 비교해볼 수 있음. (양의 비교가 아닌 비율!)


① kdeplot( , hue = 'Survived)

* common_norm = True (기본값)

```python
sns.kdeplot(x='Age', data = titanic, hue ='Survived')  # 파란색, 주황색 면적 합이 1 -> 분석하기 어려움
plt.show()
```

② kdeplot( , hue = 'Survived, common_norm = False)

```python
sns.kdeplot(x='Age', data = titanic, hue ='Survived', 
            common_norm = False)  # 전체 평균을 기준으로 그래프를 그려라. -> 서로 겹치는 부분이 평균 지점이다.!
plt.show()
```

③ multiple = 'fill' 옵션

```python
# 위의 그래프보다 더 보기 편하다. -> 전체 평균보다 생존율이 낮은지 높은지 쉽게 확인 가능
sns.kdeplot(x='Age', data = titanic, hue ='Survived'
            , multiple = 'fill') 
plt.axhline(titanic['Survived'].mean(), color = 'r')
plt.show()
```

```python
# 위의 그래프를 구간을 쪼개서 본 것.

sns.histplot(x='Age', data = titanic, bins = 16
             , hue ='Survived', multiple = 'fill')
plt.axhline(titanic['Survived'].mean(), color = 'r')
plt.show()
```

**[해석]**

```python
# 위의 그래프보다 더 보기 편하다. -> 전체 평균보다 생존율이 낮은지 높은지 쉽게 확인 가능
sns.kdeplot(x='Fare', data = titanic, hue ='Survived'
            , multiple = 'fill') 
plt.axhline(titanic['Survived'].mean(), color = 'r')
plt.show()
```

```python
# 위의 그래프를 구간을 쪼개서 본 것.

sns.histplot(x='Fare', data = titanic, bins = 16
             , hue ='Survived', multiple = 'fill')
plt.axhline(titanic['Survived'].mean(), color = 'r')
plt.show()
```
