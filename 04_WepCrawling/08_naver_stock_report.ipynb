## Crawling Naver Stock Reports
* 실습 링크: https://finance.naver.com/research/company_list.naver?&page=1

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
# 1. URL
url = 'https://finance.naver.com/research/company_list.naver?page=1'
```

```python
# 2. request(URL) > response(HTML)
response = requests.get(url)
response
```

```python
response.text[:200]  # html 문자가 잘 나오는 것을 확인
```

```python
# 3. HTML > BeautifulSoup > css-selector > DataFrame
dom = BeautifulSoup(response.content, 'html.parser')
type(dom)  # 이 안에는 select(css-selector), select_one()이 있다.
```

```python
# elements가 나오지 않으니, 아래에서 하나하나 추적하면서 파악해보기(어디에서 출력이 안되는 것인지)
selector = '#contentarea_left > div.box_type_m > table.type_1 > tbody > tr:nth-child(3)'  

# 실제 데이터와 html 개발자 도구에서 제공해주는 코드가 서로 다르다. 원래 개발자 도구가 맞는 것이다. 
# 이럴 경우, naver에서 html 도구 만든 사람이 잘못한 것이다.
selector ='#contentarea_left > div.box_type_m > table.type_1 > tbody'   # tbody에서 elements가 나오지 않는 것을 알 수 있다.

# tbody 대신 tr을 입력(사이트에서 '페이지 속성 보기'를 눌러서 데이터 소스를 확인하여 실제 존재하는 루트 입력해주기)
selector ='#contentarea_left > div.box_type_m > table.type_1 > tr'

# selector를 조금 더 간단하게 작성해보기(출력 값은 동일하다.)
selector ='table.type_1 > tr'

elements = dom.select(selector)
len(elements)
```

```python
elements[2]
```

```python
element = elements[2]
tag = element.select('td')  # td만 뽑아온 것(td는 6개)
len(tag), tag
```

```python
# 수집할 데이터 선택
data = {}

# select_one은 하나만 가져오라는 문법이다.
data['stock_name'] = tag[0].select_one('a').text    # <a class="stock_item" href="/item/main.naver?code=066570" title="LG전자">LG전자</a>
data['stock_link'] = tag[0].select_one('a').get('href')
data['title'] = tag[1].select_one('a').text
data['title_link'] = tag[1].select_one('a').get('href')
data['writer'] = tag[2].text
data['pdf_link'] = tag[3].select_one('a').get('href')
data['date'] = tag[4].text
data['pv'] = tag[5].text
data
```

```python
# enumerate
rows = []

for idx, element in enumerate(elements):
    tag = element.select('td')
    # print(idx, len(tag))
    if len(tag) == 6:   # tag가 6개인 것만 데이터 수집!
        # 수집할 데이터 선택
        data = {}
        
        # select_one은 하나만 가져오라는 문법이다.
        data['stock_name'] = tag[0].select_one('a').text    # <a class="stock_item" href="/item/main.naver?code=066570" title="LG전자">LG전자</a>
        data['stock_link'] = tag[0].select_one('a').get('href')
        data['title'] = tag[1].select_one('a').text
        data['title_link'] = tag[1].select_one('a').get('href')
        data['writer'] = tag[2].text
        data['pdf_link'] = tag[3].select_one('a').get('href')
        data['date'] = tag[4].text
        data['pv'] = tag[5].text
        rows.append(data)
```

```python
rows
```

```python
df = pd.DataFrame(rows)
df.tail(2)
```
