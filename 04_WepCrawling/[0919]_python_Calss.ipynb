```python
# 1. 클래스 선언: 코드 작성
# 예금 계좌: Account: balance, deposit(), withdraw()
class Account:
    
    balance = 0
    
    # 입금 코드 구현
    def deposit(self, amount):
        self.balance += amount
        
    # 출금 코드 구현
    def withdraw(self, amount):
        self.balance -= amount
```

```python
# 2. 객체 생성: 메모리 사용
acc1 = Account()
acc2 = Account()
```

```python
# dir(): 객체 안에 들어있는 변수들, 함수 목록 출력
dir(acc1)[-3:], dir(acc2)[-3:]
```

```python
# special methods
# - 특별한 기능을 하는 메서드
# - 앞뒤로 __가 붙은 메서드
# - 대표적으로 생성자 메서드
```

```python
# 생성자 메서드: __init__(self)
# - 특별한 기능: 객체 생성 시 실행되는 메서드
```

```python
# 1. 클래스 선언
class Account:
    # "생성자 메서드" !!: 메서드에서 사용되는 변수의 초기값 설정하는 용도 사용
    def __init__(self, balance):  
        self.balance = balance
    
    # 입금 코드 구현
    def deposit(self, amount):
        self.balance += amount
        
    # 출금 코드 구현
    def withdraw(self, amount):
        self.balance -= amount
```

```python
# 객체 생성: 메모리 사용: 원자재 사용
acc = Account(1000)  # balance 변수에 값 1000을 넣는다.(초기값 설정)
```

```python
# 메서드 호출: 불량품 생산
acc.deposit(1000)
```

```python
acc.balance
```

__add__ 메서드는 두 객체를 더할 때 호출되는 특수 메서드이다.   
하지만 여기서 "덧셈"을 구현한 것이 아니라, self.data에서 obj.data를 빼는 연산을 수행하고 있다.   
즉, num1 + num2는 num1.data에서 num2.data를 빼는 연산이 수행되고, 그 결과를 반환한다.

```python
class Number:
    def __init__(self, data):
        self.data = data
    def __add__ (self, obj):  # 인스턴스 메서드이기 때문에 객체를 만들어야 호출할 수 있다.
        self.data -= obj.data
        return self.data
```

```python
num1 = Number(3)
num2 = Number(2)
```

```python
# num1 + num2는 num1.__add__(num2)로 해석된다.
# num1은 메서드를 호출한 주체이므로 self에 대응된다.
# num2는 + 연산에서 두 번째 피연산자이므로, __add__ 메서드의 두 번째 인자인 obj에 전달된다.

# num1 .__add__(num2) : num1 객체의 __add__ 메서드 호출 num2 는 파라미터로 들어감
num1+num2  # 3-2 = 1
```
