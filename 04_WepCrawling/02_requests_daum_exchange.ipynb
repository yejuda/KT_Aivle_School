### Daum Exchange
- `https://finance.daum.net`
- status code: 4xx
- setting headers

```python
import requests
import pandas as pd
```

```python
# 1. URL
url = 'https://finance.daum.net/api/exchanges/summaries'
```

request보낼 때 user-agent 'header'값이 python이면 response할 때 403으로 에러가 뜬다.(잘못된 접근이라고 인식)   
그래서 우리는 user-agent 'header'값을 바꿔서 접근 가능하게 할 것이다.

```python
# 2. requests(url) > response(json)
response = requests.get(url)
response
```

```python
## header를 바꾸기 위해 'user-agent'와 'referer'를 지정해준 후 request진행 및 response 받아오기

# Referer 헤더는 브라우저가 이전에 방문한 페이지를 서버에 알려주는 역할
# Referer 헤더를 포함해야 하는 이유는, 서버 측에서 요청의 출처를 확인하는 보안 정책이나 사이트 관리 정책 때문이다.
# 'https://finance.daum.net/exchanges'에서 온 요청만 처리할 수 있도록 설정되어 있을 가능성이 높다.

# User-Agent 헤더는 서버에게 요청을 보내는 클라이언트(주로 브라우저나 애플리케이션)의 정보를 알려주는 역할(소프트웨어 이름, 버전, 운영 체제 정보 포함)

headers = {'user-agent':'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/129.0.0.0 Safari/537.36',
           'referer':'https://finance.daum.net/exchanges'}  
response = requests.get(url, headers = headers)
response
```

```python
response.json()
```

```python
# 3. json > list,dict > DataFrame
data = response.json()['data']
df = pd.DataFrame(data)[['currencyCode', 'currencyName', 'country', 'basePrice']]
df.tail(3)
```
